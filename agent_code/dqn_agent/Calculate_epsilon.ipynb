```python
import numpy as np
```

```python

def get_tau(N_games, eps_end):
    """
    Compute the decay factor for epsilon given the number of games to be played.
    """
    eps_fin = 0.05
    eps_init = 1.0

    N_steps = N_games * 150
    tau = - N_steps *np.log((eps_end-eps_fin) / (eps_init - eps_fin))
    return np.round(tau,0)
```

```python
get_tau(500, 0.08)
```

```python
def eric_alpha(tau):
    return(np.exp(-1/tau))
```

```python
eric_alpha(get_tau(500, 0.08))
```
